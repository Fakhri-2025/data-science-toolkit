```python
#1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("\nThis is my numpy 2x2 array:\n", array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("\nRandom 3x4 array:\n", random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("\nOnes Of Array:\n", ones_array)

print("\nZero Of Array:\n", zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)
random6x3T = random6x3.T

print("\nRandom Without Tranpose:\n", random6x3)

print("\nRandom With Tranpose:\n", random6x3T) #Tranpose
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
n, p, m = 2, 3, 4

matrix_a = np.random.rand(n, p)
matrix_b = np.random.rand(p, m)

dot_product = np.dot(matrix_a, matrix_b)

print("\nn x p:\n", matrix_a)
print("\np x m:\n", matrix_b)
print("\ndot product:\n", dot_product)
```
